```python
from pprint import pprint
from elasticsearch import Elasticsearch

es = Elasticsearch('http://localhost:9200')
client_info = es.info()
pprint('Connected to Elasticsearch successfully!')
pprint(client_info.body)
```

```python
es.indices.delete(index ='my_index', ignore_unavailable=True)
es.indices.create(index = 'my_index')
```

#### index the document sequentially

```python
import json
from tqdm import tqdm
document_ids = []
dummy_data = json.load(open("data/dummy_data.json"))
for document in tqdm(dummy_data, total=len(dummy_data)):
    response = es.index(index = 'my_index', body=document)
    document_ids.append(response['_id'])
```

```python
from pprint import pprint

response = es.update(
    index ='my_index',
    id = document_ids[0],
    script ={
        'source' : 'ctx._source.title = params.title',
        'params' : {
                 'title' : 'New Title'
             }
    },
)
pprint(response.body)
```

```python
response = es.update(
    index ='my_index',
    id = document_ids[0],
    script ={
        'source' : "ctx._source.new_field = 'test_value'",
    },
)
pprint(response.body)
```

```python
response = es.delete(index = "my_index", id = document_ids[2])
pprint(response.body)
```

#### we executed each operation one at a time, with each action requiring a separate API call.This approach is slow and inefficient. Now, let's see how to accomplish the same task using the bulk API

## Bulk API

```python
es.indices.delete(index = 'my_index', ignore_unavailable= True)
es.indices.create(index = 'my_index')
```

```python
response = es.bulk(
    operations=[
        # Action 1
        {
            "index": {
                "_index": "my_index",
                "_id": "1"
            }
        },
        # Source 1
        {
            "title": "Title 1",
            "text": "Description 1",
            "created_on": "2026-05-01"
        },
        # Action 2
        {
            "index": {
                "_index": "my_index",
                "_id": "2"
            }
        },
        # Source 2
        {
            "title": "Title 2",
            "text": "Description 2",
            "created_on": "2026-05-02"
        },
        # Action 3
        {
            "index": {
                "_index": "my_index",
                "_id": "3"
            }
        },
        # Source 3
        {
            "title": "Title 3",
            "text": "Description 3",
            "created_on": "2026-05-03"
        },
        # Action 4
        {
            "update": {
                "_id": "1",
                "_index": "my_index"
            }
        },
        # Source 4
        {
            "doc": {
                "title": "New Title"
            }
        },
        # Action 5
        {
            "update": {
                "_id": "2",
                "_index": "my_index"
            }
        },
        # Source 5
        {
            "doc": {
                "new_field": "test_value"
            }
        },
        # Action 6
        {
            "delete": {
                "_index": "my_index",
                "_id": "3"
            }
        },
    ],
)

pprint(response.body)
```

#### If errors is False, it means teh bulk API successfully executed all the actions.
